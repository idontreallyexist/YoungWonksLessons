```python
from datasets import load_dataset

ds = load_dataset("Helsinki-NLP/opus_books", "en-fr")
```

```python
print(ds['train']['translation'][0])
```

```python
import pandas as pd
```

```python
df = pd.DataFrame()
dict = {}
en=[]
fr=[]
for i in ds['train']['translation']:
    en.append(i['en'])
    fr.append(i['fr'])
dict['en']=en
dict['fr']=fr
```

```python
df=pd.DataFrame(dict)
```

```python
df.to_csv("freRaw.txt")
```

```python
with open("freRaw.txt", "r", encoding="cp1252", errors="ignore") as f:
    f.readline()
    with open("fre.txt", "w") as f2:
        for line in f:
            try:
                text=line.strip()
                x = text.split('"')
                if len(x)==4:
                    f2.write(x[1]+'\t'+x[3]+'\n')
                else:
                    x=text.split(",")
                    if len(x)==3:
                        f2.write(x[1]+'\t'+x[2]+'\n')
            except Exception as e:
                print(line)
                print(e)
```
